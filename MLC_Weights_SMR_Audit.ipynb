## Arithmetic demonstration only
For a made-up student, a released 30/40 score in a 20% category with 100 total category points contributes 6 final points and has course weight 8%. With final total 80.75 and an explicit 0.75 extra-credit contribution, recover the normalized remaining-work score and recombine. This is a test example, not an empirical result from the cohort.

```python
known_points, known_weight, extra_credit = 6., .08, .75
remaining = remaining_target(80.75, extra_credit, known_points, known_weight)
recombined = combine_final(known_points, known_weight, remaining, extra_credit)
display({'hypothetical_remaining_percentage': float(remaining), 'recombined_final_percentage': float(recombined)})
assert abs(float(recombined)-80.75) < 1e-10
```
